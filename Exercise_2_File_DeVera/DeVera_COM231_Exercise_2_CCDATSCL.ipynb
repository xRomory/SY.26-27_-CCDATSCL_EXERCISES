**24. Merging and concatenating (2 pts)**

Given:
```python
students = pd.DataFrame({"name": ["Ana", "Ben", "Carla", "Dan"],
                         "section": ["A", "B", "A", "B"]})
grades   = pd.DataFrame({"name": ["Ana", "Carla", "Dan", "Ella"],
                         "grade": [90, 85, 88, 92]})
```
- (a) Merge `students` and `grades` using an **inner** join. Print the result.
- (b) Merge `students` and `grades` using a **left** join. Print the result and add a comment explaining why Ben has `NaN`.
- (c) Use `pd.concat()` to add a new row for `"Ella"` in section `"A"` to `students`. Print the result.

````python
students = pd.DataFrame({"name": ["Ana", "Ben", "Carla", "Dan"], "section": ["A", "B", "A", "B"]})
grades   = pd.DataFrame({"name": ["Ana", "Carla", "Dan", "Ella"], "grade": [90, 85, 88, 92]})

# (a) Inner join
inner_result = pd.merge(students, grades, on='name', how='inner')
print("Inner join:")
print(inner_result)

# (b) Left join
left_result = pd.merge(students, grades, on='name', how='left')
print("\nLeft join:")
print(left_result)
# Ben has NaN because he is in students but NOT in grades.
# In a left join, all rows from the left DataFrame (students) are kept.
# Since Ben has no matching 'name' in grades, the grade column shows NaN.

# (c) Add Ella as a new row in students using pd.concat()
ella_row = pd.DataFrame({"name": ["Ella"], "section": ["A"]})
students_new = pd.concat([students, ella_row], ignore_index=True)
print("\nStudents after adding Ella:")
print(students_new)
````

- (c) Create a table named `STUDENTS` with these columns, then print `"Table created successfully"`:

| Column | Type | Rule |
|---|---|---|
| `ID` | `INT` | `PRIMARY KEY`, `NOT NULL` |
| `NAME` | `TEXT` | `NOT NULL` |
| `AGE` | `INT` | `NOT NULL` |
| `COURSE` | `CHAR(50)` | |

````python
# Write your code here
import sqlite3

# (a) Connect to the database
conn = sqlite3.connect('school.db')
print("Opened database successfully")

# (b) Run this line first so the cell can be re-run without errors
conn.execute('DROP TABLE IF EXISTS STUDENTS')

# (c) Create the STUDENTS table
conn.execute('''CREATE TABLE STUDENTS (
    ID      INT     PRIMARY KEY NOT NULL,
    NAME    TEXT    NOT NULL,
    AGE     INT     NOT NULL,
    COURSE  CHAR(50),
    GWA     REAL
);''')
print("Table created successfully")
````
